```python
import pandas as pd
import numpy as np

# Load the dataset
red_wine = pd.read_csv('winequality-red[1].csv')
print(red_wine.shape)
red_wine.head()
```

```python
red_wine['Good_Quality'] = [1 if x >= 7 else 0 for x in red_wine['quality']]
```

```python
red_wine['Good_Quality'].value_counts()
```

```python
red_wine.head(10)
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score


# Specify features and target variables

X_wine = red_wine.drop(['quality', 'Good_Quality'], axis=1)
y = red_wine['Good_Quality']

# Split the data into training and testing sets (80% train, 20% test)
X_wine_train, X_wine_test, y_train, y_test = train_test_split(X_wine, y, test_size=0.25,random_state=42)
print("Shape of X_wine_train:", X_wine_train.shape)
print("Shape of X_wine_test:", X_wine_test.shape)

# Logistic regression model
model_wine = LogisticRegression()
model_wine.fit(X_wine_train, y_train)

# Predictions
y_pred_wine = model_wine.predict(X_wine_test)

# Calculate the accuracy
accuracy = accuracy_score(y_test, y_pred_wine)
print("Accuracy score:", accuracy)
```
