```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# dataset
painting_data = pd.read_csv('monet.csv')

# Task 1: 

# Model 1:Size
size_independent_variable = painting_data['WIDTH'].values.reshape(-1, 1)

# Price
price_dependent_variable = painting_data['PRICE'].values

# data splitting
size_train_data, size_test_data, price_train_data, price_test_data = train_test_split(
    size_independent_variable, price_dependent_variable, test_size=0.2, random_state=42)

size_model = LinearRegression()
size_model.fit(size_train_data, price_train_data)

predicted_price_size = size_model.predict(size_test_data)

plt.scatter(size_test_data, price_test_data, color='blue', label='Actual Data')
plt.plot(size_test_data, predicted_price_size, color='red', label='Regression Line')
plt.title('Size vs Price')
plt.xlabel('Size')
plt.ylabel('Price')
plt.legend()
plt.show()

# error
error_size_model = mean_squared_error(price_test_data, predicted_price_size)
print("Mean Squared Error for Size model:", error_size_model)


# Model 2:Width
width_independent_variable = painting_data['WIDTH'].values.reshape(-1, 1)

# Price
price_dependent_variable = painting_data['PRICE'].values

# data Split 
width_train_data, width_test_data, price_train_data, price_test_data = train_test_split(
    width_independent_variable, price_dependent_variable, test_size=0.2, random_state=42)

width_model = LinearRegression()
width_model.fit(width_train_data, price_train_data)

predicted_price_width = width_model.predict(width_test_data)

plt.scatter(width_test_data, price_test_data, color='blue', label='Actual Data')
plt.plot(width_test_data, predicted_price_width, color='red', label='Regression Line')
plt.title('Width vs Price')
plt.xlabel('Width')
plt.ylabel('Price')
plt.legend()
plt.show()

# error
error_width_model = mean_squared_error(price_test_data, predicted_price_width)
print("Mean Squared Error for Width model:", error_width_model)


# Task 2: Multivariate Linear Regression 

multivariate_independent_variables = painting_data[['WIDTH', 'HEIGHT']]

# Price
price_dependent_variable = painting_data['PRICE'].values

multivariate_train_data, multivariate_test_data, price_train_data, price_test_data = train_test_split(
    multivariate_independent_variables, price_dependent_variable, test_size=0.2, random_state=42)


multivariate_model = LinearRegression()
multivariate_model.fit(multivariate_train_data, price_train_data)

predicted_price_multivariate = multivariate_model.predict(multivariate_test_data)


error_multivariate_model = mean_squared_error(price_test_data, predicted_price_multivariate)
print("Mean Squared Error for Multivariate model:", error_multivariate_model)
```
