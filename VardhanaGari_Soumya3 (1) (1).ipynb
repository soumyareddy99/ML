```python
import matplotlib.pyplot as plt

# Preliminary visualization
# Selected features for visualization
sel_features = ['Age_mean', 'Albumin_max', 'ALSFRS_slope']
als_train_cluster = als_train[sel_features]
plt.figure(figsize=(12, 8))
plt.scatter(als_train_cluster['Age_mean'], als_train_cluster['ALSFRS_slope'], c='blue', label='Age_mean vs ALSFRS_slope')
plt.xlabel('Age mean')
plt.ylabel('ALSFRS slope')
plt.title('Preliminary Visualization of Selected Features')
plt.grid(True)
plt.show()


```

```python
import seaborn as sns
sns.pairplot(als_train_cluster)
```

```python
from sklearn.cluster import KMeans

# Selected features for clustering
cluster_features= als_train[sel_features]

# Train k-Means models
k_val = [1, 3, 5]
k_Means_models = []
for k in k_val:
    k_means = KMeans(n_clusters=k, random_state=42)
    k_means.fit(cluster_features)
    k_Means_models.append(k_means)

# Visualize clusters for each k value
plt.figure(figsize=(12, 8))
for i, k in enumerate(k_val):
    plt.subplot(1, len(k_val), i+1)
    plt.scatter(cluster_features[sel_features[0]], cluster_features[sel_features[1]], c=k_Means_models[i].labels_, cmap='viridis')
    plt.scatter(k_Means_models[i].cluster_centers_[:, 0], k_Means_models[i].cluster_centers_[:, 1], s=300, c='red', marker='X', label='Center of Clusters')
    plt.xlabel(sel_features[0])
    plt.ylabel(sel_features[1])
    plt.title("k-Means with k={}".format(k))
    plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Evaluate the model performance
for i, k in enumerate(k_val):
    print("Center of clusters for k={}".format(k))
    print(pd.DataFrame(k_Means_models[i].cluster_centers_, columns=sel_features))
    print()

```

```python
# Visualize final clustering
i = 0
final_k = k_val[i]
plt.scatter(cluster_features[sel_features[0]], cluster_features[sel_features[1]], c=k_Means_models[i].labels_, cmap='viridis')
plt.scatter(k_Means_models[i].cluster_centers_[:, 0], k_Means_models[i].cluster_centers_[:, 1], s=300, c='red', marker='X', label='Center of clusters')
plt.xlabel(sel_features[0])
plt.ylabel(sel_features[1])
plt.title("Final Clustering Result with k={}".format(final_k))
plt.legend()
plt.show()
```
